```python
import os

base = "/content/drive/MyDrive"

targets = [
    "256x256_diffusion_uncond.pt",
    "lsun_adm.pth",
    "compute_dire.py",
    "demo.py",
    "test.jpg"
]

print("=== KIỂM TRA TÀI NGUYÊN DIRE TRÊN DRIVE ===")

found = {name: [] for name in targets}

for root, dirs, files in os.walk(base):
    dirs[:] = [
        d for d in dirs
        if d not in [".Trash", ".cache"]
    ]

    for name in targets:
        if name in files:
            path = os.path.join(root, name)
            size_mb = os.path.getsize(path) / (1024 ** 2)
            found[name].append(path)
            print(f"\n{name}")
            print(f"  Đường dẫn: {path}")
            print(f"  Dung lượng: {size_mb:.2f} MB")

print("\n=== TỔNG KẾT ===")
for name, paths in found.items():
    print(f"{name}: {'Đã tìm thấy' if paths else 'Chưa tìm thấy'}")
```
